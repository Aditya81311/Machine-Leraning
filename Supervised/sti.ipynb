```python
import pandas as pd 
import numpy as np
```

```python
data = pd.read_csv("/mnt/Machine-Leraning/Datasets/commands.csv")
```

```python
data.head()
```

```python
commands = data["command"]
actions = data["action"]
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.pipeline import Pipeline
```

```python
vectorizer = TfidfVectorizer(analyzer="char_wb",
    ngram_range=(2, 4), 
    min_df=1,)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
log_reg = LogisticRegression()
```

```python
X = vectorizer.fit_transform(commands)
```

```python
log_reg.fit(X,actions)
```

```python
import joblib
joblib.dump(vectorizer,"vectorizer.joblib")
joblib.dump(log_reg,"STI_Classifier.joblib")
print("Exported Sucessfuly")
```

```python
test_com = X[0]
```

```python
comm = ["tur on lite"]
new_com = vectorizer.transform(comm)
```

```python
log_reg.predict(new_com)
```

```python
from sklearn.preprocessing import LabelEncoder
encode = LabelEncoder()
y = encode.fit_transform(actions)
```

```python
from sklearn.model_selection import cross_val_score
```

```python
scores = cross_val_score(log_reg,X,y,cv = 5,scoring="accuracy")
```
