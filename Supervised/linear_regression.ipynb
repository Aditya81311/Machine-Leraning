## Linear Regression using 
## sklearn

```python
from sklearn.linear_model import LinearRegression
import numpy as np 
import pandas as pd 

data = pd.read_csv("data.csv")
predict_data = data[data["Type"] == "Predict"]
data = data[data["Type"]=="Train"]
X = data[["X1","X2"]].values
y = data[["y"]].values
y_pred = predict_data[["X1","X2"]].values
                                                              
```

```python
model = LinearRegression()
model.fit(X,y)

prediction = model.predict(y_pred)
prediction
```

## Cross-Validation

```python
from sklearn.linear_model import LinearRegression
import numpy as np 
import pandas as pd 

data = pd.read_csv("data.csv")
predict_data = data[data["Type"] == "Predict"]
data = data[data["Type"]=="Train"]
X = data[["X1","X2"]].values
y = data[["y"]].values
y_pred = predict_data[["X1","X2"]].values
                                             
```

```python
model = LinearRegression()
from sklearn.model_selection import cross_val_score
scores = cross_val_score(model,X,y,cv=6)
print("Cross Validation Score:: ",scores)
print("Cross Validation Score Mean:: ",scores.mean())
```
